```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import os

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    RocCurveDisplay
)

print("All libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
```

```python

zip_path = next(
    (name for name in uploaded.keys() if name.lower().endswith(".zip")),
    None
)

if zip_path is None:
    raise ValueError("Please upload your online+retail.zip dataset.")

extract_dir = "/content/online_retail"
os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_dir)

excel_files = []
for root, dirs, filenames in os.walk(extract_dir):
    for filename in filenames:
        if filename.lower().endswith((".xlsx", ".xls")):
            excel_files.append(os.path.join(root, filename))

if not excel_files:
    raise FileNotFoundError("No Excel dataset was found inside the ZIP file.")

df = pd.read_excel(excel_files[0])

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python

df = df.drop_duplicates()
df = df.dropna(subset=["CustomerID"])


df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]

df["CustomerID"] = df["CustomerID"].astype(int)
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Revenue"] = df["Quantity"] * df["UnitPrice"]


reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)

customer_df = df.groupby("CustomerID").agg(
    LastPurchase=("InvoiceDate", "max"),
    Frequency=("InvoiceNo", "nunique"),
    AvgQuantity=("Quantity", "mean"),
    AvgUnitPrice=("UnitPrice", "mean"),
    ProductVariety=("StockCode", "nunique"),
    TotalSpend=("Revenue", "sum"),
    Country=("Country", lambda x: x.mode().iloc[0])
).reset_index()

customer_df["Recency"] = (
    reference_date - customer_df["LastPurchase"]
).dt.days

customer_df = customer_df.drop(columns=["LastPurchase"])


spending_threshold = customer_df["TotalSpend"].quantile(0.75)
customer_df["CustomerType"] = (
    customer_df["TotalSpend"] >= spending_threshold
).astype(int)

print("Customer-level dataset created!")
print("Customer dataset shape:", customer_df.shape)
print("High-value spending threshold:", round(spending_threshold, 2))
print(customer_df["CustomerType"].value_counts())
customer_df.head()
```

```python

X = customer_df.drop(
    columns=[
        "CustomerID",
        "TotalSpend",
        "CustomerType"
    ]
)
y = customer_df["CustomerType"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

numeric_cols = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    exclude=np.number
).columns.tolist()

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
])

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
print("Preprocessing setup completed!")
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

# Redefine the preprocessor to output a dense matrix (sparse_output=False)
preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(sparse_output=False, handle_unknown="ignore"), categorical_cols)
])

# Set up the pipeline with the dense preprocessor
linear_model = Pipeline([
    ("preprocessing", preprocessor),
    ("pca", PCA(n_components=0.95)),
    ("svm", SVC(
        kernel="linear",
        C=1.0,
        class_weight="balanced"
    ))
])

# Fit and predict
linear_model.fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)

print("Linear SVM trained successfully!")
```

```python
# Evaluate the Linear SVM model
print("=== Linear SVM Classification Report ===")
print(classification_report(y_test, y_pred_linear))

# Calculate metrics
accuracy = accuracy_score(y_test, y_pred_linear)
precision = precision_score(y_test, y_pred_linear)
recall = recall_score(y_test, y_pred_linear)
f1 = f1_score(y_test, y_pred_linear)

print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")


fig, ax = plt.subplots(1, 2, figsize=(12, 5))


cm = confusion_matrix(y_test, y_pred_linear)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax[0])
ax[0].set_title("Confusion Matrix")
ax[0].set_xlabel("Predicted")
ax[0].set_ylabel("Actual")

RocCurveDisplay.from_estimator(linear_model, X_test, y_test, ax=ax[1])
ax[1].set_title("ROC Curve")
ax[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
nonlinear_model = Pipeline([
    ("preprocessing", preprocessor),
    ("pca", PCA(n_components=0.95)),
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        class_weight="balanced"
    ))
])

nonlinear_model.fit(X_train, y_train)

y_pred_nonlinear = nonlinear_model.predict(X_test)

print("Non-Linear SVM trained successfully!")
```

```python
print("NON-LINEAR SVM RESULTS (RBF)")

print("Accuracy:",
      round(accuracy_score(y_test, y_pred_nonlinear), 4))

print("Precision:",
      round(precision_score(y_test, y_pred_nonlinear,
                            zero_division=0), 4))

print("Recall:",
      round(recall_score(y_test, y_pred_nonlinear,
                         zero_division=0), 4))

print("F1 Score:",
      round(f1_score(y_test, y_pred_nonlinear,
                     zero_division=0), 4))

print("ROC-AUC:",
      round(roc_auc_score(
          y_test,
          nonlinear_model.decision_function(X_test)
      ), 4))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_nonlinear,
    target_names=["Regular-value", "High-value"],
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred_nonlinear)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Regular-value", "High-value"],
    yticklabels=["Regular-value", "High-value"]
)

plt.title("Non-Linear SVM (RBF) - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

RocCurveDisplay.from_estimator(
    nonlinear_model, X_test, y_test
)
plt.title("Non-Linear SVM (RBF) - ROC Curve")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear SVM", "Non-Linear SVM (RBF)"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_linear),
        accuracy_score(y_test, y_pred_nonlinear)
    ],
    "Precision": [
        precision_score(y_test, y_pred_linear, zero_division=0),
        precision_score(y_test, y_pred_nonlinear, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, y_pred_linear, zero_division=0),
        recall_score(y_test, y_pred_nonlinear, zero_division=0)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred_linear, zero_division=0),
        f1_score(y_test, y_pred_nonlinear, zero_division=0)
    ]
})

print(comparison.round(4).to_string(index=False))

comparison.set_index("Model").plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Linear vs Non-Linear SVM")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
